```python
import pandas as pd
import numpy as np

# CONFIG
PESO_PAO = 0.05
PAES_POR_TELA = 25

df = pd.read_excel(r"C:\Users\Edgar.souza\Downloads\2e26f00e-d9e9-42de-975d-3a5699a51260.xlsx")

df['Data'] = pd.to_datetime(df['Data'])

# Pegar só o início do intervalo (antes do " - ")
df['Hora_inicio'] = df['Faixa Hora N1'].str.split(' - ').str[0]
# Converter para datetime (só hora)
df['Hora'] = pd.to_datetime(df['Hora_inicio'], format='%H:%M')



dias_pt = {
    'Monday': 'Segunda-feira',
    'Tuesday': 'Terça-feira',
    'Wednesday': 'Quarta-feira',
    'Thursday': 'Quinta-feira',
    'Friday': 'Sexta-feira',
    'Saturday': 'Sábado',
    'Sunday': 'Domingo'
}

df['Dia_Semana'] = df['Data'].dt.day_name().map(dias_pt)

df['Faixa_30min'] = df['Hora'].dt.floor('30min').dt.strftime('%H:%M')




agrupado = df.groupby(
    ['Unidade Resumo', 'Dia_Semana', 'Faixa_30min']
)['Qtde. Venda'].mean().reset_index()

# Conversões
agrupado['Qtd_Paes'] = np.ceil(agrupado['Qtde. Venda'] / PESO_PAO)
agrupado['Qtd_Telas'] = np.ceil(agrupado['Qtd_Paes'] / PAES_POR_TELA)

# Gerar Excel por loja
for loja in agrupado['Unidade Resumo'].unique():
    dados_loja = agrupado[agrupado['Unidade Resumo'] == loja]
    dados_loja.to_excel(f"{loja}.xlsx", index=False)

print("Arquivos Excel gerados!")
```
