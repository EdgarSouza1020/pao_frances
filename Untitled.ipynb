```python
import pandas as pd
import numpy as np
from reportlab.platypus import SimpleDocTemplate, Table, TableStyle, Paragraph, Spacer
from reportlab.lib import colors
from reportlab.lib.styles import getSampleStyleSheet

# =========================
# CONFIGURAÇÕES
# =========================
ARQUIVO_ENTRADA = "vendas_pao.xlsx"
PESO_PAO_KG = 0.05  # 50g por pão
PAES_POR_TELA = 25
MARGEM_SEGURANCA = 1.10  # 10% a mais para evitar falta

# =========================
# LEITURA E TRATAMENTO
# =========================

df = pd.read_excel(ARQUIVO_ENTRADA)

# Converter colunas

df['Data'] = pd.to_datetime(df['Data'])
df['Hora'] = pd.to_datetime(df['Hora da venda'])

# Criar dia da semana

df['Dia_Semana'] = df['Data'].dt.day_name()

# Criar faixa de 30 minutos

df['Faixa_30min'] = df['Hora'].dt.floor('30min')

# =========================
# AGREGAÇÃO
# =========================

agrupado = df.groupby(
    ['Loja', 'Dia_Semana', 'Faixa_30min']
)['Qtde. Venda em kg'].mean().reset_index()

# =========================
# CÁLCULOS
# =========================

# Aplicar margem de segurança
agrupado['Qtde. Venda em kg'] *= MARGEM_SEGURANCA

# Converter para pães e telas
agrupado['Qtd_Paes'] = agrupado['Qtde. Venda em kg'] / PESO_PAO_KG
agrupado['Qtd_Telas'] = agrupado['Qtd_Paes'] / PAES_POR_TELA

# Arredondar para cima
agrupado['Qtd_Paes'] = np.ceil(agrupado['Qtd_Paes'])
agrupado['Qtd_Telas'] = np.ceil(agrupado['Qtd_Telas'])

# =========================
# FUNÇÃO PARA GERAR PDF
# =========================

def gerar_pdf(loja, dados_loja):
    nome_arquivo = f"producao_{loja}.pdf"
    doc = SimpleDocTemplate(nome_arquivo)
    elementos = []

    styles = getSampleStyleSheet()

    # Título
    titulo = Paragraph(f"Planejamento de Produção - {loja}", styles['Title'])
    elementos.append(titulo)
    elementos.append(Spacer(1, 12))

    # Cabeçalho da tabela
    tabela_dados = [['Dia', 'Hora', 'Kg Previsto', 'Pães', 'Telas']]

    # Ordenar dados
    dados_loja = dados_loja.sort_values(by=['Dia_Semana', 'Faixa_30min'])

    # Preencher tabela
    for _, row in dados_loja.iterrows():
        tabela_dados.append([
            row['Dia_Semana'],
            row['Faixa_30min'].strftime('%H:%M'),
            round(row['Qtde. Venda em kg'], 2),
            int(row['Qtd_Paes']),
            int(row['Qtd_Telas'])
        ])

    tabela = Table(tabela_dados, repeatRows=1)

    tabela.setStyle(TableStyle([
        ('BACKGROUND', (0, 0), (-1, 0), colors.grey),
        ('TEXTCOLOR', (0, 0), (-1, 0), colors.whitesmoke),
        ('ALIGN', (0, 0), (-1, -1), 'CENTER'),
        ('GRID', (0, 0), (-1, -1), 1, colors.black),
        ('FONTNAME', (0, 0), (-1, 0), 'Helvetica-Bold')
    ]))

    elementos.append(tabela)

    # Rodapé
    elementos.append(Spacer(1, 12))
    rodape = Paragraph(
        "* Valores com margem de segurança aplicada para evitar ruptura.",
        styles['Normal']
    )
    elementos.append(rodape)

    doc.build(elementos)

# =========================
# GERAR PDF PARA CADA LOJA
# =========================

for loja in agrupado['Loja'].unique():
    dados_loja = agrupado[agrupado['Loja'] == loja]
    gerar_pdf(loja, dados_loja)

print("PDFs gerados com sucesso!")
```
